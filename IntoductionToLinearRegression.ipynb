<h3><b> Linear regression :</b></h3>

* Linear Regression is a supervised machine learning algorithm used to predict continuous values.

* Linear Regression models the relationship between input features and output using a "best-fit" straight line.

* Linear Regression predicts continuous outputs (numerical values) such as prices, scores or values.

* Linear regression learns the relationship between independent variables (features) and the dependent variable (output) from training data.


<h3><b> Types of Linear regression :</b></h3>

* **Simple Linear Regression** : Predicts the dependent variable (the output) using a single independent variable (feature).
* **Multiple Linear Regression** : Predicts the dependent variable (The output) using 2 or more independent variables (features).

## Simple Linear Regression

* Simple Linear Regression (SLR) is a supervised learning algorithm used to **predict a continous target variable using 1 input feature**.
* Simple Linear Regression assumes a linear relationship between the input and the output and represents that linear relationship in a straight line.

```python
class SimpleLinearRegression:
    def __init__(self):
        self.weight_ = None
        self.bias_ = None
        self.r2score = None
```

* Explanation of `__init__` :

    - It is the constructor , so Python runs it automatically when you create an object of the class.
    - It is never called manually , writing `SimpleLinearRegression()` triggers it automatically.
    - Its job is to the object's attributes.

* Why do we use the underscore in `__init__` and in the attributes like `wight_` `bias_` `r2score_`? 

    - **`__init__`** needs 2 underscores on each side, it is a way to make Python recognise `__init__` as the constructor and runs it automatically when the object is created.
    - **The trailing _** on attributes is **not enforced by Python** like `__init__` , it is only a naming convention of attributes in Python.

* What is self ?
- `self` means **this particular object**.
- It is always the first parameter of a method inside a class.
- You do not pass it. Python passes it automatically.
